# Coffee Shop Sales Analysis

###  Data Analysis Project


## 1. Introduction

This project analyzes sales data from a small coffee shop.
The purpose of the analysis is to understand sales performance,
identify popular products, and determine which categories and
days generate the most revenue.

## 2. Objective

The main objective of this project is to analyze coffee shop
sales data and identify patterns that can help the business
understand its sales performance and make better decisions.

### Research Questions

1. What is the total revenue?
2. Which product sells the most?
3. Which product generates the most revenue?
4. Which category generates the most revenue?
5. Which day has the highest revenue?

## 3. Dataset Description

The dataset contains sales records from a small coffee shop.
Each record includes the date, product, category, quantity sold,
and price of the product.

### Variables

- Date: The date of the sale
- Product: The name of the product sold
- Category: The product category
- Quantity: Number of items sold
- Price: Price of one item

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 4. Import Libraries

Pandas is used to organize and analyze the data, while Matplotlib
is used to create visualizations.

load and create dataset
the code below shows the data in a table

```python
data = {
    "Date": ["Jan 1", "Jan 1", "Jan 1", "Jan 2", "Jan 2",
             "Jan 2", "Jan 3", "Jan 3", "Jan 3", "Jan 4",
             "Jan 4", "Jan 4", "Jan 5", "Jan 5", "Jan 5"],

    "Product": ["Latte", "Croissant", "Americano", "Latte", "Muffin",
                "Americano", "Latte", "Croissant", "Cappuccino",
                "Americano", "Muffin", "Latte", "Cappuccino", "Latte",
                "Croissant"],

    "Category": ["Coffee", "Food", "Coffee", "Coffee", "Food",
                 "Coffee", "Coffee", "Food", "Coffee", "Coffee",
                 "Food", "Coffee", "Coffee", "Coffee", "Food"],

    "Quantity": [3, 2, 4, 5, 3, 2, 6, 3, 4, 5, 2, 4, 3, 7, 4],

    "Price": [15, 12, 10, 15, 8, 10, 15, 12, 14, 10, 8, 15, 14, 15, 12]
}

df = pd.DataFrame(data)

df
```

## 7. Data Preparation

Revenue is calculated by multiplying the quantity sold by the
price of each product.

now we need to create new variable " Revenue"

```python
df["Revenue"] = df["Quantity"] * df["Price"]

df
```

Now your dataset has variables:

Date | Product | Category | Quantity | Price | Revenue

## 7. Data Analysis
### 7.1 What is the Total Revenue?

The total revenue is calculated by adding the revenue from all
sales records.

```python
total_revenue = df["Revenue"].sum()

total_revenue
```

The total revenue generated during the analyzed period was
731 SAR.

### 7.2 Which Product Sells the Most?

The total quantity sold for each product is calculated to identify
the best-selling product.

```python
product_sales = df.groupby("Product")["Quantity"].sum()

product_sales
```

the code below shows the best selling product

```python
product_sales.idxmax()
```

The analysis shows that [latte] was the best-selling product
based on the total quantity sold.

### 7.3 Which Product Generates the Most Revenue?

The total revenue for each product is compared to determine which
product contributes the most to sales revenue.

```python
product_revenue = df.groupby("Product")["Revenue"].sum()

product_revenue
```

the code below shows the product generating the most money

```python
product_revenue.idxmax()
```

### 7.4 Which Category Generates the Most Revenue?

Revenue is grouped by category to compare the performance of
coffee and food products.

```python
category_revenue = df.groupby("Category")["Revenue"].sum()

category_revenue
```

```python
the code below shows the category generating the most money
```

```python
category_revenue.idxmax()
```

### 7.5 Which Day Has the Highest Revenue?

Daily revenue is calculated to identify the highest-performing
sales day.

```python
daily_revenue = df.groupby("Date")["Revenue"].sum()

daily_revenue
```

the below code shows the highest performing sales day

```python
daily_revenue.idxmax()
```

## 8. Data Visualization

Visualizations are used to present the results clearly and make
patterns in the sales data easier to understand.

Now we create graphs :)

graph1 : product sales

```python
product_sales.plot(kind="bar")

plt.title("Quantity Sold by Product")
plt.xlabel("Product")
plt.ylabel("Quantity Sold")

plt.show()
```

The chart shows that [latte] had the highest number of units sold.

graph2 : Revenue by category

```python
category_revenue.plot(kind="bar")

plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue (SAR)")

plt.show()
```

in this chart it shows that coffe is the most demanding category

graph3

```python
daily_revenue.plot(kind="line", marker="o")

plt.title("Daily Revenue")
plt.xlabel("Date")
plt.ylabel("Revenue (SAR)")

plt.show()
```

in the above chart it shows that jan5th had the highest performing sales day

## 9. Findings

The analysis produced the following findings:

- The total revenue during the analyzed period was [731] SAR.
- [latte] was the best-selling product by quantity.
- [latte] generated the highest revenue.
- [coffee] generated the highest revenue among the categories.
- [jan5] recorded the highest daily revenue.
